```python
import cv2
from IPython.display import Image, display
from io import BytesIO
```

```python
image_path = "data/debug/optical-flow-marker-tracker/frame_00005_preprocessed_raw.jpg"
cv2img = cv2.imread(image_path)

image = cv2.threshold(cv2img, 170, 255, cv2.THRESH_BINARY)[1]
width = 160  # int(image.shape[1] * scale)
height = 160  # int(image.shape[0] * scale)
image = cv2.resize(image, (width, height), interpolation=cv2.INTER_LINEAR)
border_size = int(width / 8)
image = cv2.copyMakeBorder(
    image,
    top=border_size,
    bottom=border_size,
    left=border_size,
    right=border_size,
    borderType=cv2.BORDER_CONSTANT,
    value=(255, 255, 255),
)

parameters = cv2.aruco.DetectorParameters()
dictionary = cv2.aruco.getPredefinedDictionary(cv2.aruco.DICT_ARUCO_ORIGINAL)
detector = cv2.aruco.ArucoDetector(dictionary, parameters)
dmwc = detector.detectMarkersWithConfidence(image=image)
(
    detected_marker_corners,
    detected_marker_ids,
    rejected_marker_candidates,
    confidence_output,
) = dmwc
print(detected_marker_corners)
image = cv2.polylines(
    image,
    [corners.astype(int) for corners in detected_marker_corners],
    isClosed=True,
    color=(0, 255, 0),
    thickness=2,
)

io_image = BytesIO()
_, buffer = cv2.imencode(".png", image)
io_image.write(buffer)
display_image = Image(data=io_image.getvalue())
display(display_image)

print(detected_marker_ids)
print(confidence_output)
```
